Combine ttb and artsci atlas calendar scrape knowledge to produce a final output.

```python
import json
```

```python
FINAL_SAVE_PATH = "../data/courses.json"
```

```python
with open("data/courses.json", "r") as f:
    cal_courses = json.load(f)
with open("data/ttb_courses.json", "r") as f:
    ttb_courses = json.load(f)
```

```python
delete_stack = []
course_codes = []
for i, course in enumerate(cal_courses):
    # print(course)
    cc = course["course_code"]
    if cc not in ttb_courses:
        delete_stack.append(i)
    else:
        course["class_size"] = max(ttb_courses[cc]["capacityByInstructType"].values())
        course_codes.append(cc)

while len(delete_stack) > 0:
    cal_courses.pop(delete_stack.pop())

count_cal_missing = 0
for key in ttb_courses.keys():
    if key not in course_codes:
        count_cal_missing += 1
print("in ttb but missing from cal", count_cal_missing)
```

```python
# clean unknown / redundant prerequisites
from courses.course_parser import CourseParser
parser = CourseParser()
parser.courses = cal_courses
parser.clean_requisites()
```

```python
with open(FINAL_SAVE_PATH, "w", encoding="utf-8") as f:
    json.dump(parser.courses, f, indent=2)
```
